```python
from pathlib import Path

from kaggle_template.train import train_and_submit

DATA_DIR = Path("../data/raw")
TARGET = "TARGET"  # TODO: replace with the competition label column
TASK = "classification"  # or "regression"
SEED = 42
FOLDS = 5
```

```python
# Example alternate scorer; choose one compatible with the competition metric.
score = train_and_submit(
    train_path=DATA_DIR / "train.csv",
    test_path=DATA_DIR / "test.csv",
    target=TARGET,
    task=TASK,
    sample_submission_path=(DATA_DIR / "sample_submission.csv")
        if (DATA_DIR / "sample_submission.csv").exists() else None,
    output_path=Path("../submissions/experiment.csv"),
    model_path=Path("../models/experiment.joblib"),
    id_column=None,
    n_splits=FOLDS,
    seed=SEED,
    scoring=None,  # e.g. "roc_auc" or "neg_mean_absolute_error"
)
```
