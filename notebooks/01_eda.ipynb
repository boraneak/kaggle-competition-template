```python
from pathlib import Path

import pandas as pd

DATA_DIR = Path("../data/raw")
train = pd.read_csv(DATA_DIR / "train.csv")
test = pd.read_csv(DATA_DIR / "test.csv")
TARGET = "TARGET"  # TODO: replace with the competition label column

print(f"train: {train.shape}; test: {test.shape}")
display(train.head())
display(train.dtypes.to_frame("dtype"))
```

```python
missing = pd.DataFrame({
    "train_missing_pct": train.isna().mean().mul(100),
    "test_missing_pct": test.reindex(columns=train.columns.drop(TARGET), fill_value=pd.NA).isna().mean().mul(100),
})
display(missing.sort_values("train_missing_pct", ascending=False).head(30))
print("Duplicate training rows:", train.duplicated().sum())
display(train[TARGET].value_counts(dropna=False).head(20))
```
