# 05 — Submission checks

Inspect the competition's sample submission and evaluation rules before uploading. The starter writes predictions in sample-submission row/column order when that file is provided. Confirm ID alignment, prediction semantics (class labels vs probabilities), allowed ranges, and the exact expected filename/format.

```python
from pathlib import Path

import numpy as np
import pandas as pd

DATA_DIR = Path("../data/raw")
submission_path = Path("../submissions/submission.csv")  # adjust to your output file
test = pd.read_csv(DATA_DIR / "test.csv")
submission = pd.read_csv(submission_path)
sample_path = DATA_DIR / "sample_submission.csv"
sample = pd.read_csv(sample_path) if sample_path.exists() else None

assert len(submission) == len(test), "Submission row count must match test data"
assert not submission.isna().any().any(), "Submission contains missing values"
if sample is not None:
    assert submission.columns.tolist() == sample.columns.tolist(), "Columns/order differ from sample"
for column in submission.select_dtypes(include="number"):
    assert np.isfinite(submission[column]).all(), f"Non-finite values in {column!r}"

display(submission.head())
print("Submission checks passed:", submission.shape)
```
